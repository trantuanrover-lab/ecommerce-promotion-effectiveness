```python
import sqlite3
import numpy as np
import pandas as pd
```

```python
db_path = r"C:\Users\Admin\Desktop\DA Project\E-Commerce Sales Analytics\E-Commerce Sales.db"
con = sqlite3.connect(db_path)
df = pd.read_sql("SELECT * FROM fact_order_promotion_analysis", con)
con.close()
```

```python
print("Số dòng ban đầu:", len(df))
```

```python
for c in df.select_dtypes(include=["object", "string"]).columns:
    df[c] = df[c].astype("string").str.strip() 
```

```python
na = df.isna().sum()
print("Cột thiếu dữ liệu:\n", na[na > 0] if na.any() else "Không có")
```

```python
df = df.drop_duplicates() 
```

```python
print("Số dòng sau khi bỏ trùng:", len(df))
```

```python
m = df["margin_clean_pct"]
df["margin_flag"] = np.select([m.isna(), df["profit_clean"] < 0, m < 20], ["Unknown", "Loss", "Thin"], default="OK")
```

```python
df["dq_note"] = np.where(df.duplicated(["order_id", "product_id"], keep=False), "repeat_product_in_order", "OK")
```

```python
df.to_csv("fact_order_promotion_clean.csv", index=False, encoding="utf-8-sig")
```
